```python
import audio as A
import files as IO
import plotly.express as px
import numpy as np
import pandas as pd
import viz as V
```

```python
files  = IO.get_files('../data/input/')
```

```python
data = []
start = 9
end = start + 1
for f in files[start:end]:
# for i, f in enumerate(files):
    d = A.AudioData(f, 0, 0)
    d.set_spectrum()
    d.analyze()
    d.create_harmonics()
    d.resynthesize_cycle()
    data.append(d)
```

```python
d.compression_level = 5
d.create_harmonics()
d.pad_harmonics(-1)
d.resynthesize_cycle()
```

```python
V.generate_master_plot(data)
```

```python
IO.export_wavetable(data[0].frame,"../data/output",d.name)
```
